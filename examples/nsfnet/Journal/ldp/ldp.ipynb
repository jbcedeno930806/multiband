```python
# General imports
from netsim.netSimPy import *
from netsim.netSimPy.common.evaluators import NetworkEvaluator
from netsim.netSimPy.common.allocators import least_demand_percentage
```

```python
M_LAMBDA = 200000
allocator = least_demand_percentage(3)

network = Network(
    networkFileName="../../../../networks/nsfnet/network.json",
    pathsFileName="../../../../networks/nsfnet/routes.json",
    bitrateFilename="../../../../networks/nsfnet/bitrates_4_bands.json",
)
generator = EventsGenerator(mLambda=M_LAMBDA)

sim_args1 = dict(
    eventsGenerator=generator,
    network=network,
    allocator=allocator,
)
sim = NetworkSimulator(**sim_args1)


N_EVALUATIONS = 10000
traffics = [50000, 100000, 150000, 200000, 250000, 300000, 350000, 400000]

for traffic in traffics:
    callback = NetworkEvaluator(f"srmar_{int(traffic/100)}")
    sim.reset()
    sim.setLambda(traffic)
    print(f"Running simulation for srmar with lambda: {int(traffic/100)}")
    sim.run(N_EVALUATIONS, callback)
    print("\n")
```
